<h1> Step 1: Install Git </h1>


Install Git:

    MacOS:
    
        brew install git

        brew install gh [idk if this is nessicary]


    

<h1> Step 2: Configure Git </h1>

    git config --global user.name "your username"
    git config --global user.email "your.email@umn.edu"

    

<h1> Step 3: Set-up SSH Key </h1>

<h3> Step 3.1: Generate Key </h3>

    Method 1: ed25519

        ssh-keygen -t ed25519 -C "your.email@umn.edu"

    Method 2: rsa

        ssh-keygen -t rsa -b 4096 -C "your.email@umn.edu"

    
    For file location default if fine unless you want to move it somewhere.

    For the passphrase I leave it blank, but you can add one if you so desire.

<h3> Step 3.2: Start the SSH agent </h3>

    eval "$(ssh-agent -s)"

    [Replace with a different location if you moved it somewhere else when generating the key]

    ssh-add ~/.ssh/id_ed25519


<h3> Step 3.3 Copy the SSH key </h3>

    Again replace with a different location if you moved yours

    MacOS:

        cat ~/.ssh/id_ed25519.pub | pbcopy

    Linux:

        cat ~/.ssh/id_ed25519.pub

    Windows (git Bash):

        cat ~/.ssh/id_ed25519.pub

<h3> Step 3.4 Add the Key to Github </h3>

    go to github and sign-in
    
    Go to Settings -> "SSH and GPG keys" -> "New SSH key"

    Give it a title and paste the Key into the Key textbox press add ssh key


<h1> Step 4 Test the SSH connection </h1>

    ssh -T git@github.com
