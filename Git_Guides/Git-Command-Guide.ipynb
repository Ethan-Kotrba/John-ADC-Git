<h1>Push Commands</h1>

- git add .
- git commit -m "insert comment here"
- git push

<h1>Pull Commands</h1>

- git pull

<h1>Branch Commands</h1>

<h3>Switch Branches</h3>

- git checkout branch_name

<h3>Make&Switch to New Branch</h3>

- git checkout -b new_branch_name

<h3>list out branch names</h3>

- git branch

<h5> If you can't see all branches</h5>


<h7> Method 1 (Try to update local with remote)</h7>

- git fetch --all

<h7> Method 2 (shows all branches)</h7>

- git branch -a
